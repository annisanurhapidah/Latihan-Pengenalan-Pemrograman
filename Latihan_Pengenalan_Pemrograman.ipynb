```python
# ===== DATA DIRI =====
nama = "Annisa Nur Hapidah"
asal_sekolah = "SMK Negeri 1 Cimahi"
jurusan = "Rekayasa Perangkat Lunak"
alamat = "Margaasih"
pekerjaan = "IT Support dan Admin"

hobi = ["Membaca Buku", "Mendengarkan Musik", "Berlari"]
cita_cita = "Data Analyst"

# ===== BIODATA (dictionary) =====
biodata = {
    "nama": nama,
    "sekolah": asal_sekolah,
    "jurusan": jurusan,
    "alamat": alamat,
    "pekerjaan": pekerjaan,
    "hobi": hobi,
    "cita_cita": cita_cita,
}

# ===== FUNGSI PERKENALAN =====
def perkenalan(data):
    print("=" * 45)
    print("            PERKENALAN DIRI")
    print("=" * 45)
    print(f"Halo, nama saya {data['nama']}.")
    print(f"Saya bersekolah di {data['sekolah']}, jurusan {data['jurusan']}.")
    print(f"Saya tinggal di {data['alamat']} dan bekerja sebagai {data['pekerjaan']}.")
    print("\nHobi saya:")
    for i, h in enumerate(data["hobi"], start=1):
        print(f"  {i}. {h}")
    print(f"\nCita-cita saya menjadi {data['cita_cita']}.")
    print("Terima kasih sudah membaca!")
    print("=" * 45)

# ===== JALANKAN =====
perkenalan(biodata)
```
