```python
# PROGRAM KONVERSI NILAI AKADEMIK
# Preprocessing Transkrip Nilai

# 1. INPUT DAN PEMISAHAN DATA
data = input("Masukkan data (Nama, NIM, Nilai Tugas, Nilai UTS, Nilai UAS): ")
nama, nim, nilai_tugas, nilai_uts, nilai_uas = data.split(",")

# Bersihkan spasi di awal/akhir teks agar tabel rapi
nama = nama.strip()
nim = nim.strip()

# 2. KONVERSI NILAI KE FLOAT
nilai_tugas = float(nilai_tugas)
nilai_uts = float(nilai_uts)
nilai_uas = float(nilai_uas)

# 3. HITUNG NILAI AKHIR
bobot_tugas = 0.20
bobot_uts = 0.35
bobot_uas = 0.45

nilai_akhir = (nilai_tugas * bobot_tugas) + (nilai_uts * bobot_uts) + (nilai_uas * bobot_uas)

# 4. TAMPILKAN TABEL RINGKASAN
garis = "-" * 42

print(garis)
print(f"|{'RINGKASAN TRANSKRIP NILAI':^40}|")
print(garis)
print(f"| {'Nama':<14}: {nama:<23}|")
print(f"| {'NIM':<14}: {nim:<23}|")
print(garis)
print(f"| {'Komponen':<14}{'Nilai':>12}{'Bobot':>12} |")
print(f"| {'Nilai Tugas':<14}{nilai_tugas:>12.1f}{bobot_tugas:>12.0%} |")
print(f"| {'Nilai UTS':<14}{nilai_uts:>12.1f}{bobot_uts:>12.0%} |")
print(f"| {'Nilai UAS':<14}{nilai_uas:>12.1f}{bobot_uas:>12.0%} |")
print(garis)
print(f"| {'NILAI AKHIR':<14}{nilai_akhir:>12.1f}{'':>12} |")
print(garis)
```
