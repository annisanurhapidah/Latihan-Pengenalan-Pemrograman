```python

# PROGRAM KASIR OTOMATIS
# Generator Struk Transaksi

# 1. INPUT DATA PELANGGAN
nama_pelanggan = input("Masukkan nama pelanggan: ")

total_awal = float(input("Masukkan total belanja sebelum diskon: Rp"))

persentase_diskon = float(input("Masukkan persentase diskon (%): "))


# 2. VALIDASI PERSENTASE DISKON
if persentase_diskon < 0 or persentase_diskon > 100:
    print("Input ditolak! Persentase diskon harus antara 0 sampai 100.")

else:

    # 3. MENGHITUNG NOMINAL DISKON
    nominal_diskon = total_awal * persentase_diskon / 100

    # 4. MENGHITUNG TOTAL BAYAR
    total_bayar = total_awal - nominal_diskon


    # 5. MENAMPILKAN NOTA PEMBAYARAN
    print("=" * 36)
    print(f"{'NOTA PEMBAYARAN':^36}")
    print(f"{nama_pelanggan:^36}")
    print("=" * 36)

    # 6. MENAMPILKAN RINCIAN TRANSAKSI
    label_diskon = f"Diskon ({persentase_diskon:g}%)"

    print(f"{'Total Awal':<20}{f'Rp{total_awal:,.2f}':>16}")
    print(f"{label_diskon:<20}{f'Rp{nominal_diskon:,.2f}':>16}")
    print("-" * 36)
    print(f"{'Total Bayar':<20}{f'Rp{total_bayar:,.2f}':>16}")

    print("=" * 36)
    print(f"{'TERIMA KASIH':^36}")
    print("=" * 36)


```
